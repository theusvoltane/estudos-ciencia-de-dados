# Manipulando os dados

## Criando colunas numéricas

```python
url= 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'
dados= pd.read_csv(url,sep=';')
dados = dados.fillna(0)
```

```python
dados['Valor_por_mes'] = dados['Condominio'] + dados['Valor']
dados.head()
```

```python
dados['Valor_por_ano'] = dados['Valor_por_mes'] *12 + dados['IPTU']
dados.head()
```

## Criando colunas categóricas

```python
dados['Descrição'] = dados['Tipo'] + ' em ' + dados['Bairro'] + ' com ' + dados['Quartos'].astype(str) + ' quarto(s) e ' + dados['Vagas'].astype(str) + ' vaga(s) de garagem'
dados.head()
```

```python
dados['Possui_suite'] = dados['Suites'].apply(lambda x: 'Sim' if x > 0 else 'Não')
dados.head()
```

```python
dados.head()
```

```python
dados.to_csv('dados_completos.csv', index=False, sep=';')
```
